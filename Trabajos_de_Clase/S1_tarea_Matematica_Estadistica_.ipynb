# 1. Límites

```r
# Función auxiliar para aproximar límites evaluando muy cerca del punto
calcular_limite <- function(f, x0, h = 1e-7) {
  # Evaluamos un poco antes y un poco después (límite lateral)
  val_izquierda <- f(x0 - h)
  val_derecha   <- f(x0 + h)
  return(mean(c(val_izquierda, val_derecha)))
}
```

## L1. Calcular el siguiente límite trigonométrico

```r
# L1: Limit x->0 de sin(4x)/tan(x)
f1 <- function(x) sin(4*x) / tan(x)
lim_L1 <- calcular_limite(f1, 0)
cat(sprintf("L1: Lim x->0 [sin(4x)/tan(x)] aprox = %.4f (Teórico: 4)\n", lim_L1))
```

## L2. Evaluar el límite algebraico:

```r
# L2: Limit x->3 de (x^3 - 27)/(x - 3)
f2 <- function(x) (x^3 - 27) / (x - 3)
lim_L2 <- calcular_limite(f2, 3)
cat(sprintf("L2: Lim x->3 [(x^3 - 27)/(x - 3)] aprox = %.4f (Teórico: 27)\n", lim_L2))
```

## L3. Resolver el límite:

```r
# L3: Limit x->2 de (x^3 - 6x^2 + 8x)/(x - 2)
f3 <- function(x) (x^3 - 6*x^2 + 8*x) / (x - 2)
lim_L3 <- calcular_limite(f3, 2)
cat(sprintf("L3: Lim x->2 [(x^3 - 6x^2 + 8x)/(x-2)] aprox = %.4f (Teórico: -4)\n", lim_L3))
```

# 2. CONTINUIDAD

```r
# C1: f(x) por partes. Condición hallada: a + b = -3
# Vamos a probar con un ejemplo válido: a = 1, b = -4
a_val <- 1
b_val <- -4

# Definimos la función por partes en R
f_c1 <- function(x) {
  ifelse(x < 1, x^2 - 4, a_val * x + b_val)
}

lim_izq <- f_c1(1 - 1e-7) # 1 por la izquierda
lim_der <- f_c1(1 + 1e-7) # 1 por la derecha (y valor de función)

cat("C1: Verificación para a=1, b=-4 en x=1:\n")
cat(sprintf("    Límite Izquierda: %.4f\n    Límite Derecha/Función: %.4f\n", lim_izq, lim_der))
if(abs(lim_izq - lim_der) < 1e-5) cat("    -> CONCLUSIÓN: Es Continua.\n") else cat("    -> Es Discontinua.\n")
```

```r
# C2: f(x) = |x-3| + 2 en x=3
f_c2 <- function(x) abs(x - 3) + 2
val_exacto <- f_c2(3)
lim_aprox <- calcular_limite(f_c2, 3)
cat(sprintf("C2: En x=3, Valor=%.4f, Límite=%.4f -> Continua\n", val_exacto, lim_aprox))
```

```r
# C3: f(x) = sqrt(x+2) en x=-2
# R producirá NaN si evaluamos izquierda de -2, probamos continuidad por derecha
f_c3 <- function(x) sqrt(x + 2)
val_exacto_c3 <- f_c3(-2)
lim_der_c3 <- f_c3(-2 + 1e-7) # Solo derecha
cat(sprintf("C3: En x=-2, Valor=%.4f, Límite Der=%.4f -> Continua por la derecha\n", val_exacto_c3, lim_der_c3))
```

# 3. DERIVADAS (Cálculo Simbólico usando R Base)

```r
# La función D() de R toma una expresión y una variable respecto a la cual derivar.

# D1: y = sin(t)*tan(t^2 + 1)
# R requiere que escribamos las potencias y multiplicaciones explícitamente
expr_d1 <- expression(sin(t) * tan(t^2 + 1))
deriv_d1 <- D(expr_d1, "t")
cat("D1: Derivada de sin(t)*tan(t^2 + 1):\n")
print(deriv_d1)
# Nota: R a veces no simplifica (sec^2 lo muestra como 1/cos^2).
```

```r
# D2: y = sin(3*x)^4
expr_d2 <- expression(sin(3*x)^4)
deriv_d2 <- D(expr_d2, "x")
cat("\nD2: Derivada de sin^4(3x):\n")
print(deriv_d2)
```

```r
# D3: y = cos(3*x)^2
expr_d3 <- expression(cos(3*x)^2)
deriv_d3 <- D(expr_d3, "x")
cat("\nD3: Derivada de cos^2(3x):\n")
print(deriv_d3)
```

```r
# D4: y = sin(x^3 - 2*x)
expr_d4 <- expression(sin(x^3 - 2*x))
deriv_d4 <- D(expr_d4, "x")
cat("\nD4: Derivada de sin(x^3 - 2x):\n")
print(deriv_d4)
```
